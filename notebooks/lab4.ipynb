Part 12:Performance measurement: loop vs vectorized

```python
import numpy as np
import timeit

def loop_double(values):
    result = []
    for x in values:
        result.append(x * 2)
    return result

def numpy_double(values):
    return values * 2

# Question 1: Size sweep
sizes = [1_000, 100_000, 1_000_000]
rows = []

for N in sizes:
    py_list = list(range(N))
    np_arr = np.arange(N)

    n = 100 if N == 1_000 else 20

    t_loop = timeit.timeit(
        lambda: loop_double(py_list), number=n
    ) / n

    t_np = timeit.timeit(
        lambda: numpy_double(np_arr), number=n
    ) / n

    ratio = t_loop / t_np
    rows.append((N, t_loop, t_np, ratio))

# Print results as a table
print(f"{'N':>10} {'Loop (s)':>12} {'NumPy (s)':>12} {'Ratio':>10}")

for N, t_loop, t_np, ratio in rows:
    print(f"{N:>10} {t_loop:>12.6f} {t_np:>12.6f} {ratio:>10.2f}")
```

```python
import timeit

N = 1_000_000
py_list = list(range(N))
np_arr = np.arange(N)

n = 20

# Best time out of 5 runs, divided by number of executions
t_loop = min(timeit.repeat(
    lambda: loop_double(py_list),
    number=n,
    repeat=5
)) / n

t_np = min(timeit.repeat(
    lambda: numpy_double(np_arr),
    number=n,
    repeat=5
)) / n

print("Best loop time:", round(t_loop, 6), "seconds")
print("Best NumPy time:", round(t_np, 6), "seconds")
print("Speed ratio:", round(t_loop / t_np, 2))
```

The best of several runs is a fairer figure because it reduces the influence of temporary interruptions and background processes on the timing.

```python
import time
import tracemalloc

# Time the loop version
start = time.perf_counter()
result_loop = standardise_loop(data)
loop_time = time.perf_counter() - start

# Time the NumPy version
start = time.perf_counter()
result_np = standardise_np(data)
numpy_time = time.perf_counter() - start

print(f"Loop time: {loop_time:.6f} seconds")
print(f"NumPy time: {numpy_time:.6f} seconds")
print(f"Speed ratio (loop / NumPy): {loop_time / numpy_time:.2f}x")

# Compare traced Python memory allocations
tracemalloc.start()
standardise_loop(data)
_, loop_memory = tracemalloc.get_traced_memory()
tracemalloc.stop()

tracemalloc.start()
standardise_np(data)
_, numpy_memory = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"Loop peak traced memory: {loop_memory / 1024 / 1024:.2f} MB")
print(f"NumPy peak traced memory: {numpy_memory / 1024 / 1024:.2f} MB")
```
